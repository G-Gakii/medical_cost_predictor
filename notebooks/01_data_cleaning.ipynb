### Data Cleaning Tasks

1. **Standardize names:** The name column contains inconsistent capitalization. Convert names to a consistent case while preserving the original spelling of names.

2. **Convert date columns:** The `Date of Admission` and `Date of Discharge` columns are stored as strings. Convert them to the appropriate datetime data type and check for invalid date values.

3. **Investigate duplicate records:** The dataset contains 534 duplicate rows. Inspect the duplicates to determine whether they are genuine duplicate records before removing them.

```python
cleaned_df=(
    df.assign(
        Name=lambda x :x["Name"].str.strip().str.title(),
        **{
             "Date of Admission":lambda x : pd.to_datetime(x["Date of Admission"],errors="coerce"),
             "Discharge Date" : lambda x : pd.to_datetime(x["Discharge Date"],errors="coerce")
        }
       
        
    )
    .drop_duplicates()
)
```

#### Decision: Handling Duplicate Records

Duplicate records were removed because the dataset does not contain a unique patient identifier, making it difficult to verify whether repeated records belong to the same patient. However, it is unlikely that two separate patient records would have identical values across all recorded attributes, including patient name, date of admission, date of discharge, medical condition, medication, hospital, and billing amount.

Based on this assumption, exact duplicate rows were treated as redundant records and removed to prevent potential double-counting during the analysis.

### save cleaned data frame

```python
cleaned_df.to_csv("../data/healthcare-dataset/clean_healthcare_dataset.csv",index=False)
```
